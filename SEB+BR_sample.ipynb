## Calculating SEB for 0%, 25%, 50%, 75%, and 100% SM

- `SWDN`: Downward shortwave radiaton
- `SWUP`: Upward shortwave radiaton
- `LWDN`: Downward longwave radiaton
- `LWUP`: Upward longwave radiaton
- `SFLUX_T`: Sensible heat flux
- `SFLUX_R`: Latent heat flux

- `SHF_scaled` = -`SFLUX_T` * 1004 $\text{J kg}^{-1}\text{ K}^{-1}$
- `LHF_scaled` = -`SFLUX_R` * 2.5e6 J $\text{kg}^{-1}$
- `G_storage` = `SWnet` + `LWnet` + `SHF` + `LHF`

## Bowen Ratio
B = `SHF` / `LHF`

### Ground Storage

```python
ground_storage = -(shf000 + lhf000 + swnet + lwnet)
ground_storage
```

### Surface Energy Budget @ 0% Soil Moisture Plot and CSV

```python
fig, ax = plt.subplots(1, 1, figsize=(10,4))

plt.suptitle("Surface Energy Budget @ 0% Soil Moisture")

plt.tight_layout()

ax.set_xlim(pd.Timestamp('2025-06-06-050000'), pd.Timestamp('2025-06-06-183000'))

ax.plot(swnet.time, swnet, label="Shortwave Radiation", color="#fcba03")
ax.plot(lwnet.time, lwnet, label="Longwave Radiation", color="#25a15f")
ax.plot(shf000.time, shf000, label="Sensible Heat Flux", color="#eb4034")
ax.plot(lhf000.time, lhf000, label="Latent Heat Flux", color="#18aaed")
ax.plot(ground_storage.time, ground_storage, label="Ground Storage", color="#7d4d1a")

ax.axhline(0, color="black", ls=":", zorder=0)
ax.legend(bbox_to_anchor=(1,0.5))

ax.set_xlabel("Time")
ax.set_ylabel("(W/m$^{2})$")

ax.xaxis.set_major_formatter(
    mdates.DateFormatter("%H:%M")
)

plt.tight_layout()
plt.savefig("SEB000.png", dpi=300)
plt.show()
```

```python
SEB000 = xr.Dataset()

SEB000 = SEB000.assign(shf000 = shf000)
SEB000 = SEB000.assign(lhf000 = lhf000)
SEB000 = SEB000.assign(sw = swnet)
SEB000 = SEB000.assign(lw = lwnet)
SEB000 = SEB000.assign(g = ground_storage)

SEB000 = SEB000.drop_vars(['alt', 'z'])

SEB000 = SEB000.to_dataframe()

SEB000.to_csv('SEB000.csv')
print(SEB000)
```
