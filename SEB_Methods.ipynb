# Surface Energy Budget Analysis - Methods Summary

## Step 1: Data Loading
Using the lab's `rams_tools` library, I load RAMS model output from `datapath`. The function `get_rams_filepaths()` returns a list of filepaths for a given soil moisture scenario (e.g., "sm050" for 50% soil moisture).

## Step 2: Variable Extraction
For each file, I extract:
- `SWDN`: Downwelling shortwave radiation
- `SWUP`: Upwelling shortwave radiation
- `LWDN`: Downwelling longwave radiation
- `LWUP`: Upwelling longwave radiation
- `SFLUX_T`: Sensible heat flux
- `SFLUX_R`: Latent heat flux

## Step 3: Spatial Averaging
Using `ds.sel(z=1).mean(dim=("x", "y"))`, I average each variable across the horizontal domain to produce a single time series.

## Step 4: Flux Calculations
- Net shortwave = SWDN - SWUP
- Net longwave = LWDN - LWUP
- Scaled sensible heat = -SFLUX_T × 1004
- Scaled latent heat = -SFLUX_R × 2.5e6
- Ground heat storage G = sum of all components

## Step 5: Visualization
I create time series plots using matplotlib, formatting the x-axis for hour-minute labels.
